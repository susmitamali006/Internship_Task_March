```python
# Import important libraries 
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
# Load dataset
df = pd.read_csv('adult.csv')
df
```

```python
# first five rows of datatset
df.head()
```

```python
# Last five rows of datatset
df.tail()
```

```python
# # Count of total number of Rows and Columns 
df.shape
```

```python
# Structure of dataset
df.info()
```

```python
# Statistic summary of dataset
df.describe()
```

```python
# Column names 
df.columns 
```

```python
# Standardize column names 
df.columns = (
    df.columns.str.replace('.', '_')
) 
df.columns
```

```python
# Check null values 
df.isnull().sum()
```

```python
# Check duplicates 
df.duplicated().sum()
```

```python
# remove duplicates
df.drop_duplicates(inplace=True)
```

```python
# After removing duplicates 
df.duplicated().sum()
```

```python
# count ? in columns 
print((df == '?').sum())
```

```python
# replace that ? with NaN
df.replace('?', np.nan, inplace=True)
print((df == '?').sum())
```

```python
# Fill every NaN cell with the word 'Unknown'
df.fillna('Unknown', inplace=True)
```

```python
# distribution of the income column
print("Income Category Distribution:")
print(df['income'].value_counts())
```

```python
# percentage of each group
print("\nPercentage Distribution:")
print(df['income'].value_counts(normalize=True) * 100)
```

```python
# Average age and hours by Income
print(df.groupby('income')[['age', 'hours_per_week']].mean())
```

```python
# Average Education level
print(df.groupby('income')['education_num'].mean())
```

```python
# Compare numeric columns
display(df.groupby('income')[['age', 'education_num', 'hours_per_week', 'capital_gain', 'capital_loss']].mean())
```

```python
correlation_matrix = df.corr(numeric_only=True)
display(correlation_matrix)
```

```python
# new column 
df['income_binary'] = df['income'].apply(lambda x: 1 if x == '>50K' else 0)
print("Binary Income Column:")
display(df[['income', 'income_binary']].head(10))
print("\nCorrelation with High Income (>50K):")
print(df.corr(numeric_only=True)['income_binary'].sort_values(ascending=False))
```

```python
low_income = df[df['income_binary'] == 0]['age']
high_income = df[df['income_binary'] == 1]['age']

plt.figure(figsize=(10, 6))
plt.hist(low_income, bins=20, alpha=0.5, label='<=50K', color='orange')
plt.hist(high_income, bins=20, alpha=0.5, label='>50K', color='blue')
plt.title('Age Distribution by Income Group', fontsize=14)
plt.xlabel('Age', fontsize=12)
plt.ylabel('Number of People', fontsize=12)
plt.legend(loc='upper right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# percentage of high earners (>50K) 
edu_income = df.groupby('education')['income_binary'].mean().sort_values() * 100
plt.figure(figsize=(10, 8))
plt.barh(edu_income.index, edu_income.values, color='skyblue')
plt.title('Earning >50K by Education Level', fontsize=14)
plt.xlabel('Percentage of People', fontsize=12)
plt.ylabel('Education Level', fontsize=12)
plt.axvline(24, color='red', linestyle='--', label='Average (24%)')
plt.legend()
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```

```python
df['age_group'] = pd.cut(df['age'], bins=[10, 20, 30, 40,50,60,70,80,90])
hours_by_age = df.groupby(['age_group', 'income'], observed=False)['hours_per_week'].mean().unstack()
hours_by_age.plot(kind='bar', figsize=(12, 6), color=['orange', 'blue'], edgecolor='black')
plt.title('Average Hours Worked per Week by Age Group', fontsize=14)
plt.ylabel('Average Hours', fontsize=12)
plt.xlabel('Age Group', fontsize=12)
plt.xticks(rotation=45)
plt.legend(title='Income')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
