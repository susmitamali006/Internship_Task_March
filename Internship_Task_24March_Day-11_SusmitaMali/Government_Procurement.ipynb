```python
# Import important libraries 
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Standardize column names 
df.columns = (
    df.columns 
    .str.replace('.', '_')
)
df.columns 
```

```python
# check null values 
df.isnull().sum()
```

```python
# Check duplicates
df.duplicated().sum()
```

```python
# Change date into standardize date format
df['award_date'] = pd.to_datetime(df['award_date'], dayfirst=True)
df
```

```python
# new column year 
df['year'] = df['award_date'].dt.year
df
```

```python
df['awarded_amt'] = pd.to_numeric(df['awarded_amt'], errors='coerce')
df
```

```python
# Total sum per year
yearly_spending = df.groupby('year')['awarded_amt'].sum()
```

```python
yearly_spending.plot(kind='bar', color='skyblue')
plt.title('Total Government Spending by Year')
plt.ylabel('Amount Spent')
plt.show()
```

```python
top_vendors = df['supplier_name'].value_counts().head(5)
print(top_vendors)
```

```python
top_vendors.plot(kind='barh', color='red')
plt.title('Top 5 Suppliers by Contract Count)')
plt.xlabel('Number of Contracts Won')
plt.show()
```

```python
#Top Vendor
top_vendor_data = df[df['supplier_name'] == 'NCS PTE. LTD.']
ncs_trend = top_vendor_data.groupby('year').size()
print(ncs_trend)
```

```python
# Performance Trend Analysis
ncs_trend.plot(kind='line', marker='o', color='purple')
plt.title('Performance Trend: NCS PTE. LTD.')
plt.ylabel('Number of Contracts')
plt.show()
```
