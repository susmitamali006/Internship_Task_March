```python
# For Numerical columns to handle null values
numeric_cols = df.select_dtypes(include=['number']).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())

# Object/Text columns to handle null values
object_cols = df.select_dtypes(include=['object']).columns
df[object_cols] = df[object_cols].fillna("Unknown")

print(df.isnull().sum())
```

```python
# check duplicates
df.duplicated().sum()
```

```python
# Define feature 
features = ['Income', 'Poverty', 'Professional', 'Service', 'Unemployment']
# input
X = df[features]

print("Features selected for clustering:")
X.head()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
# Scale the data
X_scaled = scaler.fit_transform(X)

print("Scaled Data:")
print(X_scaled[0])
```

```python
from sklearn.cluster import KMeans

# an empty list errors 
errors = []
for k in range(1, 11):
    # Create a "Brain" Model with k number of groups
    model = KMeans(n_clusters=k, random_state=42)
    # differenciate areas are similar and which are different compared with X_scaled
    model.fit(X_scaled)
    errors.append(model.inertia_) # inertia= Error Score

plt.plot(range(1, 11), errors, marker='o')
plt.title('The Elbow Method')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Error Rate')
plt.show()
```

```python
kmeans_model = KMeans(n_clusters=4, random_state=42)
#new column Segment
df['Segment'] = kmeans_model.fit_predict(X_scaled)
```

```python
print("Socio Economic")
summary = df.groupby('Segment')[features].mean()
print(summary.sort_values(by='Income', ascending=False))
```

```python
print(df[['State', 'County', 'Income', 'Poverty', 'Segment']].head())
```

```python
# Average Income for each Segment
analysis = df.groupby('Segment')['Income'].mean().sort_values()

analysis.plot(kind='bar', color=['red', 'orange', 'blue', 'green'], figsize=(10,6))
plt.title('Average Income by Segment')
plt.xlabel('Socio-Economic Segment (0, 1, 2, 3)')
plt.ylabel('Average Income ($)')
plt.show()
```

```python
# Assign segment 
names = {
    0: "High-Need Area",
    1: "Middle Class",
    2: "Wealthy Professional",
    3: "Working Class",    
}
df['Segment_Name'] = df['Segment'].map(names)
print("New column Segment_Name:")
df[['State', 'County', 'Income', 'Segment_Name']].head(10)
```
