```python
# Standardise column names 
df.columns = (
    df.columns 
    .str.lower()
    .str.strip()
    .str.replace(" ","_")
)
df.columns
```

```python
# There are number of Countries are present in dataset. 
# From that I selected USA
df_country = df[df['country'] == 'USA'].copy()
df_country = df_country.sort_values('year')
print(f"We have {len(df_country)} years of data for the analysis.")
```

```python
# calculate the trend for perticular country with years
df_country.set_index('year', inplace=True)
# Create the Trend column
df_country['temp_trend'] = df_country['avg_temperature_(°c)'].rolling(window=10).mean()
df_country[['avg_temperature_(°c)', 'temp_trend']].head(15)
```

```python
# last rows for USA with average temperature
df_country[['avg_temperature_(°c)', 'temp_trend']].tail(15)
```

```python
# Group the data by year to get the 'Global Average' temperature for each year
# Here we get only one data for each year (remove duplicate years) 
df_global = df.groupby('year')['avg_temperature_(°c)'].mean().reset_index()
df_global = df_global.sort_values('year')

# New Trend column for this new clean data
df_global['temp_trend'] = df_global['avg_temperature_(°c)'].rolling(window=5).mean()
df_global.head(10)
```

```python
# Plot the Actual Average Temperature 
plt.figure(figsize=(12, 6))
plt.plot(df_global['year'], df_global['avg_temperature_(°c)'], marker='o', linestyle='-', alpha=0.4, label='Yearly Average', color='blue')

# Plot the 5-Year Trend 
plt.plot(df_global['year'], df_global['temp_trend'], color='red', linewidth=3, label='5-Year Warming Trend')
plt.title('Global Temperature Trend Analysis (2000 - 2024)', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Temperature (°C)', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
#Need toCalculate the 'Baseline' (The overall average temperature of the whole period)
baseline_temp = df_global['avg_temperature_(°c)'].mean()

# Calculate the Anomaly for each year
# Anomaly = Actual Temperature - Baseline
df_global['anomaly'] = df_global['avg_temperature_(°c)'] - baseline_temp
std_dev = df_global['avg_temperature_(°c)'].std()
df_global['significant_anomaly'] = df_global['anomaly'].abs() > (1.5 * std_dev)
print(f"The Global Baseline Temperature is: {baseline_temp:.2f}°C")
df_global[['year', 'avg_temperature_(°c)', 'anomaly', 'significant_anomaly']].head(10)
```

```python
# Plot
colors = ['red' if x > 0 
          else 'blue' for x in df_global['anomaly']]
plt.figure(figsize=(12, 6))
plt.bar(df_global['year'], df_global['anomaly'], color=colors, alpha=0.7)
plt.axhline(0, color='black', linewidth=2, linestyle='--') # Baseline 
plt.title('Global Temperature Anomalies of USA(2000 - 2024)', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Baseline (°C)', fontsize=12)
plt.grid(True, axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
# Group all columns by year (taking the mean of all countries)
df_countries = df.groupby('year').agg({
    'avg_temperature_(°c)': 'mean',
    'co2_emissions_(tons/capita)': 'mean',
    'sea_level_rise_(mm)': 'mean',
    'rainfall_(mm)': 'mean'
}).reset_index()
df_countries.head(10)
```

```python
# calculate the correlation matrix
cols = ['avg_temperature_(°c)', 'co2_emissions_(tons/capita)', 
        'sea_level_rise_(mm)', 'rainfall_(mm)']
corr_matrix = df_countries[cols].corr().values # Get the numerical values
fig, ax = plt.subplots(figsize=(10, 8))
im = ax.imshow(corr_matrix, cmap='coolwarm', vmin=-1, vmax=1)
plt.colorbar(im)
ax.set_xticks(np.arange(len(cols)))
ax.set_yticks(np.arange(len(cols)))
ax.set_xticklabels(cols, rotation=45, ha="right")
ax.set_yticklabels(cols)
# correlation value for each box
for i in range(len(cols)):
    for j in range(len(cols)):
        text = ax.text(j, i, f"{corr_matrix[i, j]:.2f}",
                       ha="center", va="center", color="black", fontsize=12)

plt.title('Correlation Analysis: Climate Indicators', fontsize=14)
plt.tight_layout()
plt.show()
```
